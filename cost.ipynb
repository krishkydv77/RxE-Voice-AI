Sarvam STT Cost:
Yeh audio duration ya size par depend karta hai. Agar aapke paas audio ki length ya audio_kb hai, toh uske anusaar per KB ya per second rate set kar sakte hain:$$\text{STT Cost} = \text{audio\_kb} \times \text{rate\_per\_kb}$$Sarvam TTS Cost:
Yeh total input characters (input_chars) par depend karta hai:$$\text{TTS Cost} = \left(\frac{\text{input\_chars}}{1000}\right) \times \text{rate\_per\_1k\_chars}$$Groq LLM Cost:
Yeh prompt aur completion tokens par depend karta hai:$$\text{LLM Cost} = \left(\frac{\text{prompt\_tokens}}{1000000} \times \text{input\_token\_rate}\right) + \left(\frac{\text{completion\_tokens}}{1000000} \times \text{output\_token\_rate}\right)$$Vector DB (FastEmbed):
Kyunki yeh locally run hota hai aur iska koi API charge nahi lagta, iski cost 0 hogi.SerpAPI (Web Search):
Jab bhi web search trigger hoti hai, toh per request ek fixed cost judti hai (jaise $0.005 per search).
